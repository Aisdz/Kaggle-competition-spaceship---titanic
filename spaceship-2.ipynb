```python
train["CabinRegion"] = pd.qcut(train["CabinNum"], q=6, duplicates="drop").astype(str)

cabin_bins = np.unique(np.quantile(train["CabinNum"].dropna(), q=np.linspace(0, 1, 7)))

test["CabinRegion"] = pd.cut(
    test["CabinNum"],
    bins=cabin_bins,
    include_lowest=True
).astype(str)
```

```python
catboost_features = [
    "HomePlanet",
    "CryoSleep",
    "Destination",
    "Age",
    "VIP",
    "RoomService",
    "FoodCourt",
    "ShoppingMall",
    "Spa",
    "VRDeck",
    "Deck",
    "CabinNum",
    "Side",
    "GroupSize",
    "TotalSpend",
    "NoSpend",
    "IsAlone",
    "AgeGroup",
    "SpendPerPerson",
    "CabinGroup",
    "VIP_CryoSleep",
    "PlanetDestination",
    "log_RoomService",
    "log_FoodCourt",
    "log_ShoppingMall",
    "log_Spa",
    "log_VRDeck",
    "log_TotalSpend",
    "log_SpendPerPerson",
    "CabinRegion"
]

catboost_categorical = [
    "HomePlanet",
    "CryoSleep",
    "Destination",
    "VIP",
    "Deck",
    "Side",
    "AgeGroup",
    "CabinGroup",
    "VIP_CryoSleep",
    "PlanetDestination",
    "CabinRegion"
]

print("Total features:", len(catboost_features))
print("Categorical features:", catboost_categorical)
```

```python
# categorical features -> replace NaN with "Missing"
for col in catboost_categorical:
    train[col] = train[col].astype(object).where(train[col].notna(), "Missing")
    test[col] = test[col].astype(object).where(test[col].notna(), "Missing")

# numeric features -> fill with train median
catboost_numeric = [col for col in catboost_features if col not in catboost_categorical]

for col in catboost_numeric:
    median_value = train[col].median()
    train[col] = train[col].fillna(median_value)
    test[col] = test[col].fillna(median_value)

print("Train missing:", train[catboost_features].isnull().sum().sum())
print("Test missing:", test[catboost_features].isnull().sum().sum())
```

```python
from catboost import CatBoostClassifier
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score
from catboost import CatBoostClassifier
import numpy as np

X_cv = train[catboost_features]
y_cv = train["Transported"]

cat_features_idx = [X_cv.columns.get_loc(col) for col in catboost_categorical]

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

cv_scores = []

for fold, (train_idx, val_idx) in enumerate(cv.split(X_cv, y_cv), 1):
    X_train_fold = X_cv.iloc[train_idx]
    X_val_fold = X_cv.iloc[val_idx]
    y_train_fold = y_cv.iloc[train_idx]
    y_val_fold = y_cv.iloc[val_idx]

    cv_model = CatBoostClassifier(
        iterations=2000,
        learning_rate=0.03,
        depth=8,
        loss_function="Logloss",
        eval_metric="Accuracy",
        random_seed=42,
        verbose=0
    )

    cv_model.fit(
        X_train_fold,
        y_train_fold,
        cat_features=cat_features_idx,
        eval_set=(X_val_fold, y_val_fold),
        use_best_model=True
    )

    val_pred = cv_model.predict(X_val_fold)
    fold_score = accuracy_score(y_val_fold, val_pred)
    cv_scores.append(fold_score)

    print(f"Fold {fold} accuracy: {fold_score:.6f}")

cv_scores = np.array(cv_scores)

print("\nCV scores:", cv_scores)
print("Mean CV accuracy:", cv_scores.mean())
print("Std CV accuracy:", cv_scores.std())
```

```python
from sklearn.model_selection import train_test_split

X_cb = train[catboost_features]
y_cb = train["Transported"]

X_train_cb, X_val_cb, y_train_cb, y_val_cb = train_test_split(
    X_cb,
    y_cb,
    test_size=0.2,
    stratify=y_cb,
    random_state=42
)

print(X_train_cb.shape)
print(X_val_cb.shape)
```

```python
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

cat_features_idx = [X_train_cb.columns.get_loc(col) for col in catboost_categorical]

cat_model = CatBoostClassifier(
    iterations=2000,
    learning_rate=0.03,
    depth=8,
    loss_function="Logloss",
    eval_metric="Accuracy",
    random_seed=42,
    verbose=200
)

cat_model.fit(
    X_train_cb,
    y_train_cb,
    cat_features=cat_features_idx,
    eval_set=(X_val_cb, y_val_cb),
    use_best_model=True
)

cb_val_pred = cat_model.predict(X_val_cb)

print("Validation accuracy:", accuracy_score(y_val_cb, cb_val_pred))
print()
print(classification_report(y_val_cb, cb_val_pred))
print()
print(confusion_matrix(y_val_cb, cb_val_pred))
```

```python
X_full_cb = train[catboost_features]
y_full_cb = train["Transported"]
X_test_cb = test[catboost_features]

cat_features_idx_full = [X_full_cb.columns.get_loc(col) for col in catboost_categorical]

final_cat_model = CatBoostClassifier(
    iterations=2000,
    learning_rate=0.03,
    depth=8,
    loss_function="Logloss",
    eval_metric="Accuracy",
    random_seed=42,
    verbose=200
)

final_cat_model.fit(
    X_full_cb,
    y_full_cb,
    cat_features=cat_features_idx_full
)

test_pred_cb = final_cat_model.predict(X_test_cb)
```

```python
submission = pd.DataFrame({
    "PassengerId": test_ids,
    "Transported": test_pred_cb.astype(bool)
})

submission.to_csv("submission.csv", index=False)
submission.head()
```
